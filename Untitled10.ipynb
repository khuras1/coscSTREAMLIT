```python
nominal_cols = ['env_condition', 'visitor_preferred_language', 'visitor_country', 'event_event_type']
ordinal_cols = ['env_season', 'visitor_age_group']

print("Nominal columns:", nominal_cols)
print("Ordinal columns:", ordinal_cols)
```

## Encode nominal data

### Subtask:
Apply one-hot encoding to the nominal columns.


**Reasoning**:
Apply one-hot encoding to the nominal columns in the DataFrame.



```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')

encoded_nominal = ohe.fit_transform(data_dropped_ids[nominal_cols])
encoded_nominal_df = pd.DataFrame(encoded_nominal, columns=ohe.get_feature_names_out(nominal_cols))

data_encoded = data_dropped_ids.drop(columns=nominal_cols).reset_index(drop=True)
data_encoded = pd.concat([data_encoded, encoded_nominal_df], axis=1)

display(data_encoded.head())
```

## Encode ordinal data

### Subtask:
Apply ordinal encoding to the ordinal columns.


**Reasoning**:
Apply ordinal encoding to the ordinal columns 'env_season' and 'visitor_age_group' in the `data_encoded` DataFrame.



## Combine encoded data

### Subtask:
Concatenate the encoded columns with the remaining numerical columns.


**Reasoning**:
Concatenate the encoded nominal and ordinal columns with the remaining numerical and datetime columns from the original dataframe.



```python
# Identify non-encoded and non-ordinal columns
remaining_cols = data_dropped_ids.columns.difference(nominal_cols + ordinal_cols)
data_remaining = data_dropped_ids[remaining_cols].reset_index(drop=True)

# Combine the encoded data with the remaining columns
data_final = pd.concat([data_encoded.reset_index(drop=True), data_remaining], axis=1)
display(data_final.head())
```

*   Ordinal encoding was applied to the ordinal columns (`env_season`, `visitor_age_group`) based on predefined category orders, converting them into numerical representations in the `data_encoded` DataFrame.
*   The final `data_final` DataFrame was created by combining the `data_encoded` DataFrame (containing the encoded nominal and ordinal features) with the remaining numerical and datetime columns from the original `data_dropped_ids` DataFrame.

Evaluate the performance of the trained ensemble models on the test data (`X_test`, `y_test`) using appropriate metrics.

Import necessary evaluation metrics and calculate MAE, MSE, and R-squared for the trained ensemble models on the test data.



```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

ensemble_model_performance = {}

for model_name, model in trained_models.items():
    if model_name in ['Random Forest Regressor', 'Gradient Boosting Regressor']:
        y_pred = model.predict(X_test)

        mae = mean_absolute_error(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)

        ensemble_model_performance[model_name] = {'MAE': mae, 'MSE': mse, 'R-squared': r2}

display(ensemble_model_performance)
```
